```python
import re

def process_email(email):
    at_count = email.count('@')
    if at_count != 1 or '.' not in email.split('@')[-1]:
        return "Invalid email"
    
    username = email.split('@')[0]
    domain_part = email.split('@')[1]
    domain_name = domain_part[:domain_part.rfind('.')]
    
    if email.endswith(".com"):
        domain_type = "Commercial Domain"
    elif email.endswith(".edu"):
        domain_type = "Educational Domain"
    else:
        domain_type = "Other Domain"
        
    return f"Username: {username}, Domain: {domain_name}, Type: {domain_type}"

def decode_message(encoded_str, mapping):
    match = re.search(r'[a-zA-Z]+\s[a-zA-Z]+', encoded_str)
    if not match:
        return "Invalid format"
    
    core = match.group()
    word1, word2 = core.split()
    
    decoded_word1 = word1[::-1].capitalize()
    
    decoded_word2 = word2
    for old, new in mapping.items():
        decoded_word2 = decoded_word2.replace(old, new)
        
    return f"{decoded_word1} {decoded_word2}"

if __name__ == "__main__":
    print("Task-1:", process_email("Amit_ml@gmail.edu"))
    print("Task-2:", decode_message("###!!@mocleW EPGTQ!!!6789", {}))
    print("Task-3:", decode_message("&&&**$gnirtS PLIO!!@1234", {'I':'E', 'O':'U'}))
    print("Task-4:", decode_message("##$$$@!yalpstcejorp EPUVT****9887", {'E':'A', 'U':'O'}))
```
