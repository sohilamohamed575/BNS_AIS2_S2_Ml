```python
#Task 1 — BankAccount

class BankAccount:
    def __init__(self, initial_balance=0):
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        
        self.balance = initial_balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
            return
        
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
            return
        
        if amount > self.balance:
            print("Insufficient funds.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

    def check_balance(self):
        print(f"Current balance: {self.balance}")


# Usage
account = BankAccount(100)

account.deposit(50)
account.withdraw(30)
account.check_balance()
```

```python
#Task 2 — Calculator Class

class Calculator:
    def __init__(self):
        pass

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        if b == 0:
            return "Error! Division by zero."
        
        return a / b

    def calculator(self):
        print("Welcome to the calculator!")
        print("Choose an operation:")
        print("1: Add")
        print("2: Subtract")
        print("3: Multiply")
        print("4: Divide")

        choice = input("Enter the number of the operation you want to perform (1/2/3/4): ")

        if choice not in ['1', '2', '3', '4']:
            print("Invalid choice! Please select 1, 2, 3, or 4.")
            return

        try:
            num1 = float(input("Enter the first number: "))
            num2 = float(input("Enter the second number: "))
        except ValueError:
            print("Invalid input! Please enter numbers only.")
            return

        if choice == '1':
            print(f"{num1} + {num2} = {self.add(num1, num2)}")

        elif choice == '2':
            print(f"{num1} - {num2} = {self.subtract(num1, num2)}")

        elif choice == '3':
            print(f"{num1} * {num2} = {self.multiply(num1, num2)}")

        elif choice == '4':
            result = self.divide(num1, num2)
            print(f"{num1} / {num2} = {result}")


# Usage
calculator = Calculator()
calculator.calculator()
```

```python
#Task 3 — Abstract Animal Class

from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass

    def describe(self):
        print("This is an animal.")


class Dog(Animal):

    def make_sound(self):
        return "Woof!"


class Cat(Animal):

    def make_sound(self):
        return "Meow!"


class Cow(Animal):

    def make_sound(self):
        return "Moo!"


# Testing the classes

dog = Dog()
cat = Cat()
cow = Cow()

print("Dog:")
dog.describe()
print("Sound:", dog.make_sound())

print("\nCat:")
cat.describe()
print("Sound:", cat.make_sound())

print("\nCow:")
cow.describe()
print("Sound:", cow.make_sound())
```

```python
#Task 4 — TextFileReader

class TextFileReader:

    def __init__(self, file_path):
        self.file_path = file_path
        self.content = ""

    def read_file(self):
        try:
            with open(self.file_path, 'r', encoding='utf-8') as file:
                self.content = file.read()

            return self.content

        except FileNotFoundError:
            print("Error: File not found.")
            self.content = ""
            return ""

        except IOError:
            print("Error: Could not read the file.")
            self.content = ""
            return ""

    def count_lines(self):
        return len(self.content.splitlines())

    def count_words(self):
        return len(self.content.split())

    def count_characters(self):
        return len(self.content)

    def display_content(self):
        print(self.content)


# Usage
reader = TextFileReader("test.txt")

reader.read_file()

print("Number of lines:", reader.count_lines())
print("Number of words:", reader.count_words())
print("Number of characters:", reader.count_characters())

print("\nFile Content:")
reader.display_content()
```

```python
#Task 5 — Unique Words & Frequency

def count_word_frequency(words):
    unique_words = set(words)

    counts = {}

    for word in unique_words:
        counts[word] = 0

        for item in words:
            if word == item:
                counts[word] += 1

    return counts


# Usage
words = [
    "Welcome",
    "Ali",
    "Hi",
    "Ali",
    "No",
    "Hi",
    "No",
    "Ali",
    "No",
    "Ali"
]

result = count_word_frequency(words)

print(result)
```

```python
#Task 6 — Read File + UserExtractor

def read_txt_file(file_path):
    """Reads the contents of a text file and returns it as a string."""

    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()

        return content

    except FileNotFoundError:
        return f"Error: The file '{file_path}' was not found."

    except IOError:
        return "Error: An error occurred while reading the file."


class UserExtractor:

    def __init__(self, file_path):
        self.file_path = file_path
        self.usernames = {}

    def extract_usernames(self):
        content = read_txt_file(self.file_path)

        # Check if reading the file returned an error
        if content.startswith("Error:"):
            return content

        lines = content.splitlines()

        for line in lines:

            if ":" in line:
                username, password = line.split(":", 1)

                username = username.strip()
                password = password.strip()

                if username:
                    self.usernames[username] = password

        return self.usernames


# Usage
if __name__ == "__main__":

    file_path = "users.txt"

    extractor = UserExtractor(file_path)

    result = extractor.extract_usernames()

    print("Extracted usernames:")
    print(result)
```
