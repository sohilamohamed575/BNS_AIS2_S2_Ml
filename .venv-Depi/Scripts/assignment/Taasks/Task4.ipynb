```python
# Task 1: Add an item to a tuple

class TupleItemAdder:
    """
    Handles adding an item to a tuple.

    Attributes:
        my_tuple (tuple): The original tuple.
        item (any): The item to be added.
    """

    def __init__(self, my_tuple, item):
        self.my_tuple = my_tuple
        self.item = item

    def add_item(self):
        """
        Adds an item to the tuple.

        Returns:
            tuple: A new tuple containing the added item.
        """
        temp_list = list(self.my_tuple)
        temp_list.append(self.item)
        return tuple(temp_list)


# Create object
task1 = TupleItemAdder((1, 2, 3), 4)

# Call method and print result
print("Task 1:", task1.add_item())
```

```python
# Task 2: Sum all items in a list

class ListSummer:
    """
    Handles calculating the sum of all items in a list.

    Attributes:
        numbers (list): A list of numbers.
    """

    def __init__(self, numbers):
        self.numbers = numbers

    def calculate_sum(self):
        """
        Calculates the sum of all numbers in the list.

        Returns:
            int or float: The total sum.
        """
        total = 0

        for number in self.numbers:
            total += number

        return total


# Create object
task2 = ListSummer([5, 10, 15, 20])

# Call method and print result
print("Task 2:", task2.calculate_sum())

```

```python
# Task 3: Multiply all items in a list

class ListMultiplier:
    """
    Handles multiplying all items in a list.

    Attributes:
        numbers (list): A list of numbers.
    """

    def __init__(self, numbers):
        self.numbers = numbers

    def calculate_product(self):
        """
        Calculates the product of all numbers in the list.

        Returns:
            int or float: The multiplication result.
        """
        result = 1

        for number in self.numbers:
            result *= number

        return result


# Create object
task3 = ListMultiplier([2, 3, 4])

# Call method and print result
print("Task 3:", task3.calculate_product())
```

```python
# Task 4: Find the smallest number

class SmallestNumberFinder:
    """
    Finds the smallest number in a list without using min().

    Attributes:
        numbers (list): A list of numbers.
    """

    def __init__(self, numbers):
        self.numbers = numbers

    def find_smallest(self):
        """
        Finds the smallest number manually.

        Returns:
            int or float: The smallest number.
        """
        smallest = self.numbers[0]

        for number in self.numbers:
            if number < smallest:
                smallest = number

        return smallest


# Create object
task4 = SmallestNumberFinder([10, 3, 25, 1, 7])

# Call method and print result
print("Task 4:", task4.find_smallest())
```

```python
# Task 5: Find the largest number

class LargestNumberFinder:
    """
    Finds the largest number in a list without using max().

    Attributes:
        numbers (list): A list of numbers.
    """

    def __init__(self, numbers):
        self.numbers = numbers

    def find_largest(self):
        """
        Finds the largest number manually.

        Returns:
            int or float: The largest number.
        """
        largest = self.numbers[0]

        for number in self.numbers:
            if number > largest:
                largest = number

        return largest


# Create object
task5 = LargestNumberFinder([10, 3, 25, 1, 7])

# Call method and print result
print("Task 5:", task5.find_largest())
```

```python
# Task 6: Count strings with length >= 2

class StringLengthCounter:
    """
    Counts strings whose length is greater than or equal to 2.

    Attributes:
        words (list): A list of strings.
    """

    def __init__(self, words):
        self.words = words

    def count_valid_strings(self):
        """
        Counts strings with length >= 2.

        Returns:
            int: Number of valid strings.
        """
        count = 0

        for word in self.words:
            if len(word) >= 2:
                count += 1

        return count


# Create object
task6 = StringLengthCounter(["hi", "a", "hello", "no", "on"])

# Call method and print result
print("Task 6:", task6.count_valid_strings())
```

```python
# Task 7: Clone or copy a list

class ListCloner:
    """
    Creates a copy of a list.

    Attributes:
        original (list): The original list.
    """

    def __init__(self, original):
        self.original = original

    def clone_list(self):
        """
        Creates and returns a copy of the original list.

        Returns:
            list: A copied list.
        """
        return self.original.copy()


# Create object
task7 = ListCloner([1, 2, 3, 4])

# Call method and print result
print("Task 7:", task7.clone_list())
```

```python
# Task 8: Remove an item from a set

class SetItemRemover:
    """
    Handles removing an item from a set.

    Attributes:
        my_set (set): The original set.
        item (any): The item to remove.
    """

    def __init__(self, my_set, item):
        self.my_set = my_set
        self.item = item

    def remove_item(self):
        """
        Removes the specified item from the set.

        Returns:
            set: The updated set.
        """
        self.my_set.discard(self.item)
        return self.my_set


# Create object
task8 = SetItemRemover({1, 2, 3, 4, 5}, 3)

# Call method and print result
print("Task 8:", task8.remove_item())
```

```python
# Task 9: Check if one set is a subset of another

class SubsetChecker:
    """
    Checks whether one set is a subset of another set.

    Attributes:
        set1 (set): The set that will be checked.
        set2 (set): The set that may contain set1.
    """

    def __init__(self, set1, set2):
        self.set1 = set1
        self.set2 = set2

    def is_subset(self):
        """
        Checks if set1 is a subset of set2.

        Returns:
            bool: True if set1 is a subset, otherwise False.
        """
        return self.set1.issubset(self.set2)


# Create object
task9 = SubsetChecker({1, 2}, {1, 2, 3, 4})

# Call method and print result
print("Task 9:", task9.is_subset())
```
