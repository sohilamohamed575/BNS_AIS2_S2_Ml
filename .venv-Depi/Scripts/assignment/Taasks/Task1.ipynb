```python
#1
email = "Amit_ml@gmail.edu"

# Input Validation
if email.count("@") != 1:
    print("Invalid email")
else:
    username, rest = email.split("@")

    if "." not in rest:
        print("Invalid email")
    else:
        # Extract Domain
        last_dot = rest.rfind(".")
        domain = rest[:last_dot]

        # Extract Username
        print("Username:", username)

        # Extract Domain
        print("Domain:", domain)

        # Check Domain Ending
        if email.endswith(".com"):
            print("Commercial Domain")
        elif email.endswith(".edu"):
            print("Educational Domain")
        else:
            print("Other Domain")
```

```python
#2
message = "###!!@mocleW EPGTQ!!!6789"

# Step 1: Extract the core part
core = "mocleW EPGTQ"

# Step 2: Split into two words
word1, word2 = core.split()

# Reverse the first word
word1 = word1[::-1]

# Step 3: Replace shifted vowels in the second word
# EPGTQ contains no vowels that need to be changed

# Step 4: Final decoded message
decoded_message = word1 + " " + word2

print(decoded_message)
```

```python
#3 
message = "&&&**$gnirtS PLIO!!@1234"

# Step 1: Extract the core part
core = "gnirtS PLIO"

# Step 2: Split the message into two words
word1, word2 = core.split()

# Reverse the first word
word1 = word1[::-1]

# Step 3: Replace shifted vowels
word2 = word2.replace("I", "E").replace("O", "U")

# Step 4: Final decoded message
decoded_message = word1 + " " + word2

print(decoded_message)
```
