```python
# 1. Add an item to a tuple
my_tuple = (1, 2, 3)
item_to_add = 4
new_tuple = my_tuple + (item_to_add,)
print("1:", new_tuple)
```

```python
# 2. Sum all items in a list
numbers = [5, 10, 15, 20]
total = 0
for number in numbers:
    total += number
print("2:", total)
```

```python
# 3. Multiply all items in a list
numbers = [2, 3, 4]
result = 1
for number in numbers:
    result *= number
print("3:", result)
```

```python
# 4. Smallest number without min()
numbers = [10, 3, 25, 1, 7]
smallest = numbers[0]
for number in numbers:
    if number < smallest:
        smallest = number
print("4:", smallest)
```

```python
# 5. Largest number without max()
numbers = [10, 3, 25, 1, 7]
largest = numbers[0]
for number in numbers:
    if number > largest:
        largest = number
print("5:", largest)
```

```python
# 6. Count strings with length >= 2
words = ["hi", "a", "hello", "no", "on"]
count = 0
for word in words:
    if len(word) >= 2:
        count += 1
print("6:", count)
```

```python
# 7. Clone/copy a list
original = [1, 2, 3, 4]
cloned = original.copy()
print("7:", cloned)
```

```python
# 8. Remove an item from a set
my_set = {1, 2, 3, 4, 5}
my_set.remove(3)
print("8:", my_set)
```

```python
# 9. Check subset
set1 = {1, 2}
set2 = {1, 2, 3, 4}
print("9:", set1.issubset(set2))
```

```python
# 10. Remove all elements from a set
my_set = {1, 2, 3}
my_set.clear()
print("10:", my_set)
```

```python
# 11. Find max and min manually
my_set = {10, 3, 25, 1, 7}
values = list(my_set)
maximum = values[0]
minimum = values[0]
for value in values:
    if value > maximum:
        maximum = value
    if value < minimum:
        minimum = value
print("11:", "Max:", maximum, "Min:", minimum)
```

```python
# 12. Find index of an item in a tuple
my_tuple = (10, 20, 30, 40)
index_value = my_tuple.index(30)
print("12:", index_value)
```

```python
# 13. Convert tuple to dictionary
my_tuple = (("a", 1), ("b", 2), ("c", 3))
my_dict = dict(my_tuple)
print("13:", my_dict)
```

```python
# 14. Unzip list of tuples
pairs = [(1, 2), (3, 4), (5, 6)]
list1, list2 = zip(*pairs)
list1 = list(list1)
list2 = list(list2)
print("14:", list1, list2)
```

```python
# 15. Reverse a tuple
my_tuple = (1, 2, 3, 4, 5)
reversed_tuple = my_tuple[::-1]
print("15:", reversed_tuple)
```

```python
# 16. Convert list of tuples to dictionary
pairs = [("a", 1), ("b", 2), ("c", 3)]
my_dict = dict(pairs)
print("16:", my_dict)
```

```python
# 17. Replace last value of each tuple
data = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
new_list = []
for item in data:
    temp = list(item)
    temp[-1] = 100
    new_list.append(tuple(temp))
print("17:", new_list)
```

```python
# 18. Sort tuple by float element
items = (("item1", 10.5), ("item2", 5.7), ("item3", 8.9))
sorted_items = tuple(sorted(items, key=lambda item: item[1]))
print("18:", sorted_items)
```

```python
# Challenge 1: Longest word
words = ["hi", "hello", "fantastic", "world"]
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print("Challenge 1:", longest)
```

```python
# Challenge 2: Merge dictionaries
d1 = {"a": 1, "b": 2}
d2 = {"c": 3, "d": 4}
result = {**d1, **d2}
print("Challenge 2:", result)
```

```python
# Challenge 3: Sum dictionary values
data = {"a": 10, "b": 20, "c": 30}
total = 0
for value in data.values():
    total += value
print("Challenge 3:", total)
```

```python
# New Challenge: Sum all student scores
scores = {
    "student1": {"math": 80, "science": 90},
    "student2": {"math": 75, "science": 85},
    "student3": {"math": 92, "science": 88},
}

total = 0
for student_data in scores.values():
    for score in student_data.values():
        total += score

print("New Challenge:", total)
```
