```python
import pandas as pd
import numpy as np
```

```python
#Data 
data = {
    'Transaction_ID': range(1, 11),
    'Product_Category': ['Electronics', 'Home', 'Electronics', 'Sports', 'Home',
                         'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'],
    'Sales_Amount': [150, 200, 155, 300, 210, 180, 205, 1000, 190, 160],
    'Customer_Age': [25, 34, np.nan, 45, 23, 31, 29, np.nan, 38, 40],
    'Rating': [5, 4, 3, 5, 2, 4, 5, 2, 4, 3]
}

df_test = pd.DataFrame(data)
df_test.to_csv("company_sales_test.csv", index=False)
```

```python
# Assignment 1
def automated_stat_analyzer(df, column_name):
    """
    Return statistical information for a numerical column,
    or the mode for a categorical column.
    """
    if column_name not in df.columns:
        raise KeyError(f"Column '{column_name}' does not exist.")

    series = df[column_name]

    if pd.api.types.is_numeric_dtype(series):
        mean = series.mean()
        median = series.median()
        std = series.std()

        if mean > median:
            skewed = "Right-skewed"
        elif mean < median:
            skewed = "Left-skewed"
        else:
            skewed = "Not skewed"

        return {
            "Mean": mean,
            "Median": median,
            "Standard Deviation": std,
            "Skewed": skewed
        }

    mode = series.mode()
    return {
        "Mode": mode.iloc[0] if not mode.empty else None
    }

print("Assignment 1 - Sales_Amount:")
print(automated_stat_analyzer(df_test, "Sales_Amount"))

print("\nAssignment 1 - Product_Category:")
print(automated_stat_analyzer(df_test, "Product_Category"))
```

```python
# Assignment 2
def null_handling_strategy(df, strategy="fill_mean"):
    """
    Handle missing values using:
    - drop_rows
    - fill_mean
    - fill_median

    Mean/median are applied only to numerical columns.
    The original DataFrame is not modified.
    """
    if strategy not in {"drop_rows", "fill_mean", "fill_median"}:
        raise ValueError(
            "Invalid strategy. Use 'drop_rows', 'fill_mean', or 'fill_median'."
        )

    result = df.copy()

    if not result.isnull().values.any():
        return result

    if strategy == "drop_rows":
        return result.dropna()

    numeric_cols = result.select_dtypes(include=np.number).columns

    for col in numeric_cols:
        if strategy == "fill_mean":
            result[col] = result[col].fillna(result[col].mean())
        elif strategy == "fill_median":
            result[col] = result[col].fillna(result[col].median())

    return result

print("\nAssignment 2 - fill_mean:")
print(null_handling_strategy(df_test, "fill_mean"))

print("\nAssignment 2 - fill_median:")
print(null_handling_strategy(df_test, "fill_median"))

print("\nAssignment 2 - drop_rows:")
print(null_handling_strategy(df_test, "drop_rows"))
```
